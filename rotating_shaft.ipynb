```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
train_0 = pd.read_csv("E:/CODE\Python/Vibration_Rotating_Shaft/0D.csv")
train_4 = pd.read_csv("E:/CODE\Python/Vibration_Rotating_Shaft/4D.csv")
```

```python
train_0 = train_0.sample(frac=0.01, random_state=42)
train_4 = train_4.sample(frac=0.01, random_state=42)
```

```python
train_0['unbalancing'] = 0
train_4['unbalancing'] = 1
```

```python
train_0.info()
train_4.info()
```

```python
test_0 = pd.read_csv("E:/CODE/Python/Vibration_Rotating_Shaft/0E.csv")
test_4 = pd.read_csv("E:/CODE/Python/Vibration_Rotating_Shaft/4E.csv")
```

```python
test_0 = test_0.sample(frac=0.01, random_state=42)
test_4 = test_4.sample(frac=0.01, random_state=42)
```

```python
test_0['unbalancing'] = 0
test_4['unbalancing'] = 1
```

```python
test_0.info()
test_4.info()
```

Concat them

```python
train = pd.concat([train_0, train_4], axis=0)
test = pd.concat([test_0, test_4], axis=0)
```

Data cleaning

```python
m1 = train['Measured_RPM'].quantile(0.999)
m2 = train['Measured_RPM'].quantile(0.001)
print(m1, m2)
```

keep the RPM between upper and lower boundary

```python
train = train[(train['Measured_RPM']<m1) & (train['Measured_RPM']>m2)]
```

```python
v1 = train['Vibration_1'].quantile(0.998)
print(v1)
```

```python
train = train[(train['Vibration_1']<v1)]
```

Preprocessing

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
scaler.fit(train)
train_scaled = scaler.transform(train)
X_train = pd.DataFrame(train, index=train.index, columns=train.columns)
y_train=train['unbalancing']
```

```python
X_train.columns
```

```python
X_train = X_train[['V_in', 'Measured_RPM', 'Vibration_1', 'Vibration_2', 'Vibration_3']]
```
